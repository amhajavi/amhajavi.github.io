```python
import json 
import re
from pathlib import Path
from alive_progress import alive_bar

file = Path('../../data/source')

wavnames = [f for f in file.iterdir() if f.is_file() and f.suffix == '.wav']

wavnames = sorted(wavnames)
```

```python
def sentence_case(text):
    if not text:
        return text
    
    # normalise whitespace around punctuation
    text = re.sub(r'\s*([.!?])\s*', r'\1 ', text).strip()
    
    # punctuations -> split sentences
    sentences = [s.strip() for s in re.split(r'(?<=[.!?\s])+', text) if s.strip()]
    
    processed = []
    for sent in sentences:
        if not sent:
            continue
        
        # sentence case
        if sent:
            sent = sent[0].upper() + sent[1:]
        
        # standalone 'i' -> 'I'
        # Match ' i ' or start/end boundaries
        sent = re.sub(r'\bi\b', 'I', sent)
        
        processed.append(sent)
    
    output_text = ' '.join(processed)

    if output_text[-1] not in ["?", "!", "."]:
        output_text = output_text
    return output_text

# Test
# test_cases = [
#     "hello world. i am here. what about this? i think it works.",
#     "i went to the store. it was closed. i came home.",
#     "multiple   spaces.   i  am  testing.",
#     ""
# ]

# for t in test_cases:
#     print(f"Input:  '{t}'")
#     print(f"Output: '{sentence_case(t)}'")
#     print("-" * 40)
```

```python
titles, classes = [], []
with alive_bar(len(wavnames), title="Processing") as bar:
    for i, d in enumerate(wavnames):
        # sentence case 
        title = f"{i+1})"
        classname = sentence_case(sentence_case(d.stem.split("_")[0]))
        
        titles.append(title)
        classes.append(classname)
        bar()


with open("title.json", 'w') as f:
    json.dump(titles, f, indent=4, ensure_ascii=False)

with open("info.json", 'w') as f:
    json.dump(classes, f, indent=4, ensure_ascii=False)
```
